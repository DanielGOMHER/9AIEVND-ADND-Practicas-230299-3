*12.3Funciones que mandan a llamar otras funciones  y que el resultado  de un procesamiento, en esre caso utilizaremos una funcion predeterminada en Python denominada replace que nos permite  sustituir letras en una  cadena de texto.*

*Definicion:*

```python
def sustituir_vocales(texto):
    texto = texto.replace("A", "4")
    texto = texto.replace("a", "4")

    texto = texto.replace("E", "3")
    texto = texto.replace("e", "3")

    texto = texto.replace("I", "1")
    texto = texto.replace("i", "1")

    texto = texto.replace("O", "0")
    texto = texto.replace("o", "0")

    texto = texto.replace("U", "U")
    texto = texto.replace("u", "U")

    return texto
```

*Llamado*

```python
print(sustituir_vocales(saludar_nombre("Daniel")))
```

**18. FILTRAMOS LOS VALORES EXTREMOS DE LA COLUMNA MILES**

En este paso se eliminan los valores muy pequeños y muy grandes de la columna MILES, ya que pueden afectar el análisis y las gráficas.

**23. GRAFICA DE CLASIFICACIÓN DE VIAJES POR CATEGORIA**

En este paso se realizó una gráfica de clasificación de viajes por categoría para comparar la cantidad de registros pertenecientes a cada tipo de viaje. La gráfica representa que la categoría Business contiene la mayor cantidad de viajes registrados, mientras que la categoría Personal tiene una cantidad considerablemente menor.

```python
fig, ax  = mat.subplots(figsize=(8,5))
grafica= sea.countplot(
    data=df,
    x='CATEGORY',
    hue='CATEGORY',
    palette= 'Set2',
    legend=False,
    ax=ax)
#Titulo
ax.set_title('Categorias')
#Agregar etiquetas con totales
for container in ax.containers:
    ax.bar_label(container)
    
mat.show()
```

**24. GRAFICA DE VIAJES POR PROPOSITO (PURPOSE)**


Se realizó una gráfica de viajes por propósito (PURPOSE) para analizar cuáles son los motivos más frecuentes de los viajes registrados en el conjunto de datos. La gráfica representa que la categoría Others es la que concentra la mayor cantidad de viajes, seguida por propósitos como Meeting y Meal/Entertain. En cambio, categorías como Charity ($), Commute y Airport/Travel presentan muy pocos registros.

```python
fig, ax = mat.subplots(figsize=(10,6))

grafica = sea.countplot(
    data=df,
    y='PURPOSE',
    hue='PURPOSE',
    dodge= False,
    palette='viridis',
    legend=False
)
#Titulo
ax.set_title('Purpose of Visit (Others = Not Mentioned)')

#Agregar Totales
for container in ax.containers:
    ax.bar_label(container)

mat.show()
```

**25. GRAFICA DE LUGAR DE ABORDAJE (START)**

En este paso se realizó una gráfica de los lugares de abordaje (START) para identificar cuáles son los puntos desde donde se realizan más viajes dentro del conjunto de datos. La gráfica representa la cantidad de viajes iniciados en cada lugar de abordaje.

```python
fig, ax = mat.subplots(figsize=(15,50))

#Ordenar alfabeticamente
order=sorted(df['START'].dropna().unique())

grafica = sea.countplot(
    data=df,
    y='START',
    order=order,
    hue='START',
    dodge= False,
    palette='viridis',
    legend=False,
    ax=ax
)
#Titulo
ax.set_title('Lugar de Aborde')

#Agregar Totales
for container in ax.containers:
    ax.bar_label(container)

mat.show()
```
